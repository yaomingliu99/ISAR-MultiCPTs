```python
from nutbolt import sar_cptlin_runner
from nutbolt.metadata import DEFAULT_METADATA, MetadataRegistry
```

```python
data_folder = Path("/mnt/d/Research/SAR With Python/Nultbolt_2027/Jul_2026/datasets/processed")
# Working folder
master_folder = Path("/mnt/d/Research/SAR With Python/Nultbolt_2027/Jul_2026/Multiple CPTs")
master_folder.mkdir(parents=True, exist_ok=True)
os.chdir(master_folder)
```

```python
df = pd.read_parquet(data_folder / "sar5yr_for_cptlin.parquet")
```

```python
df.shape
```

```python
df.columns = df.columns.str.lower()
```

```python
output_folder = master_folder
```

## Start modeling for All cases Morbidity

```python
model_name="Morbidity_pymc"
print(f"\n{"*"*40}\n{model_name}\n{"*"*40}\n")
sar_cptlin_runner.run_model(
    outcome="score1b",
    model_name=model_name,
    cluster_var="prncptx",
    df=df.copy(),
    predictors = ["agegroup_bucket", "asaclas", "fnstatpressurg"],
    target_accept=0.3,
    chains_per_run=4,
    n_total_chains=4,
    tune=1000,
    draws=3000,
    low_rank_modified_mass_matrix=True,
    mass_matrix_eigval_cutoff=20,
    mass_matrix_gamma =1e-2,
    tolerance=1000,
    check_every_s=20,
    max_time_s=3600, 
    seed=123,
    output_folder=output_folder)
```
